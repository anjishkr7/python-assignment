### **`Problem-2:`** Write a Python function that accepts a hyphen-separated sequence of words as parameter and returns the words in a hyphen-separated sequence after sorting them alphabetically.

**Example 1:**

Input:
```bash
green-red-yellow-black-white
```

Output:
```bash
black-green-red-white-yellow
```

````python
# Write code here
def sortingAlphabetically(chr):
  new_list=chr.split('-')
  new_list.sort()
  return '-'.join(new_list)
a=sortingAlphabetically("green-red-yellow-black-white")
print(a)
````

### **`Problem 3:`** Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

```
Sample String : 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
Expected Output :
No. of Upper case characters :  9
No. of Lower case Characters :  47
```

````python
# Write code here
s='CampusX is an Online Mentorship Program fOr EnginEering studentS.'
def upLow(string):
  l=[]
  upCount=0
  lowCount=0
  for i in string:
    if i.isupper():
      upCount+=1
    elif i.islower():
      lowCount+=1
  l.extend([upCount,lowCount])
  return l
a=upLow(s)
print(f"No. of Upper case characters : {a[0]}\nNo. of Lower case characters : {a[1]}")
````

### **`Problem 5:`** Write a Python function to check whether a number is perfect or not.

A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

```
The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6.
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6.

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.
```


````python
# Write code here
def isPerfectNum(n):
  divisors_arr=[]
  for i in range(1,n+1):
    if n%i==0:
      divisors_arr.append(i)
  if int(sum(divisors_arr)/2)==n:
    return f"Its a Perfect Number: {n}"
  else:
    return "Its not a perfect number"
a=isPerfectNum(28)
print(a)
````

`Problem-8` Write a python function that receives a list of integers and prints out a histogram of bin size 10

```
Input:
[13,42,15,37,22,39,41,50]
```

```
Output:
{11-20:2,21-30:1,31-40:2,41-50:3}
```

````python
# Write code here
def makeHisto(l):
  histogram = {
    "11-20": 0,
    "21-30": 0,
    "31-40": 0,
    "41-50": 0
  }
  for i in l:
    if i in range(11,21):
      histogram['11-20']+=1
    elif i in range(21,31):
      histogram["21-30"]+=1
    elif i in range(31,41):
      histogram["31-40"]+=1
    elif i in range(41,51):
      histogram["41-50"]+=1
  return histogram
print(makeHisto([13,42,15,37,22,39,41,50]))




````

```
Query Point
(0,0)
```

```
Output

```

````python
# Write code here
import sys
def closest_coordinate(l,q):
  min_dist=sys.maxsize
  result=()
  for i in l:
    a=(((i[0]-q[0])**2+(i[1]-q[1])**2)**(1/2))
    if a<min_dist:
      min_dist=a
      result=(i[0],i[1])
  return result
list_2d=[(1,1),(2,2),(3,3),(4,4)]
query_pt=(0,0)
res=closest_coordinate(list_2d,query_pt)
print(f"Nearest to {query_pt} is {res}")
````

`Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings

https://en.wikipedia.org/wiki/Bag-of-words_model

````python
# Write code here
l=["hello world", "hello python", "world python"]
def bagOpr(l):
  word_count_dict={}
  for i in l:
    for j in i.split():
      if j in word_count_dict:
        word_count_dict[j]+=1
      else:
        word_count_dict[j]=1
  return word_count_dict

print(bagOpr(l))
````
