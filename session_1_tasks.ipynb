### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))

distance = ((x2 - x1) ** 2 + (y2 - y1) ** 2) ** 0.5

print("Euclidean Distance:", distance)
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
# Write your code here
principle=int(input("Enter the principle: "))
rate_of_interest=int(input("Enter the rate of interest: "))
time_period=int(input("Enter the time period: "))
simple_interest=(principle*rate_of_interest*time_period)/100
print("Simple Interest is:",simple_interest)
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads=int(input("Enter the number of Heads: "))
legs=int(input("Enter the number of Legs: "))
dog=int((legs-2*heads)/2)
chicken=int(heads-dog )
print("DOGS:",dog,"Chicken:",chicken)
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here
num=int(input("Enter the number: "))
sum=0
for i in range(1,num+1):
  sum+=(i**2)
print(sum)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
a=int(input("Enter the 1st term: "))
b=int(input("Enter the 2st term: "))
n=int(input("Enter the postion of the term which u want to find: "))
common_diff=b-a
nth_term=a+((n-1)*common_diff)
print(nth_term)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
#volume of milk tank
#dimensions of milk tank
H=int(input("Enter the height of the tank: "))
L=int(input("Enter the length of the tank: "))
B=int(input("Enter the breadth of the tank: "))
volume_of_the_tank=H*L*B
#dimensions of glass
h=int(input("Enter the height of glass: "))
r=int(input("Enter the radius of glass: "))
volume_of_glass=3.14*(r**2)*h
glass_of_milk=volume_of_the_tank/volume_of_glass
print(int(glass_of_milk))
````
